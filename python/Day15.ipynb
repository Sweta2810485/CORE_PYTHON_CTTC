```python
try:
    a = 10/0
    print(a)
except:
    print("Dhang se lekh")
```

```python
try:
    a = 10/0
    print(a)
except Exception as e:
    print(f" you have {e}")
```

one try block multiple exceptional block


```python
try:
    x= int(input("Enter a value"))
    a = 10/x
    print(a)
except ZeroDivisionError:
    print("Give me the right value:")
except ValueError:
    print("Value error")
```

```python
import traceback
try:
    a = 10/0
    print(a)
except Exception as e:
    print("You have",e)
    traceback.print_exc()
```
