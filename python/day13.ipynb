```python
# create a class named Rectangle with attributes length and width, and methods to calculate area and perimeter.


class Rectangle:
    length = 10
    width = 5

    def display(self):
        area = self.length * self.width
        perimeter = 2 * (self.length + self.width)
        print(f"Length: {self.length}")
        print(f"Width: {self.width}")
        print(f"Area: {area}")
        print(f"Perimeter: {perimeter}")
```

```python
rect = Rectangle()
rect.display()
```

```python
class car:
    company = "Tata"
    def __init__(self,brand,model,fuel_percent,km_driven):
        self.brand = brand
        self.model = model
        self.fuel= fuel_percent
        self.km = km_driven
    def display(self):
        print(f"Car Brand:{self.brand} \nCar Model:{self.model} \nCar fuel_percent:{self.fuel} \nCar km:{self.km}")
    def fill(self,amt):
        self.fuel +=amt
        print(f"Updated fuel : {self.fuel}")
    def drive(self,km_amt):
        self.km +=km_amt
        self.fuel -= 10
        print(f"Updated km: {self.km} \nupdated fuel: {self.fuel}")
```

```python
c = car("tiago","b4",45,32456)
c.display()
c.fill(26)
c.drive(23)
```
