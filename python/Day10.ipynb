```python
def check_password(p):
    up = 0
    lo = 0
    di = 0
    space = 0
    sp = 0
    for i in p:
        if i.isupper():
            up +=1
        elif i.islower():
            lo +=1
        elif i.isdigit():
            di +=1
        elif i.isspace():
            space +=1
        elif i.isalnum()==False:
            sp +=1
    if up>0 and lo>0 and di>0 and space==0 and sp>0 and len(p)>8:
        print("valid password")
    else:
        print("Invalid")
```

```python
check_password("Rajendra@123")
```

```python
password = input("Enter a name")
check_password(password)
```
