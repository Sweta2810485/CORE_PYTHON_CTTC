```python
# 2- default parametre
# you can assign a default value to a parameter. if no value is passed, the default is used. If you pass a value, it overrdes the default. right to left
def show(a= 0, b= 0,c=0,d=0):
    print(a,b,c,d)
show(5,"hi")
show(1,3,4,5.3)
show()
```

```python
# 3- keyword parameter
# you password  the argument by name, not by position. This makes your code more readable and the order of argument doesn't matter.
def show(a= 0, b= 3,c=10,d=0):
    print(a,b,c,d)
show(1,2,3,4.3)
show(d=23,c=65,a=43,b=45)
show(c=30)
```

```python
# 4- variable length parameter
# *args (Variable_length positional Argument)
# The *args syntax allows a function to accept any number of positional argument. inside the function, args is trated as a tuple.

def show(*a):
    print(a)
show(10,12.34)
show("hi",1,2.3)
```

```python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result
```

```python
# calling with a varying number of argument
print(multiply(2,3))
print(multiply(2,3,4))
print(multiply(1,2,3,4,5))
```

```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(key,':' , value)
```

```python
# calling with varying keyword argument
print_info(name = "Alice",age =30,city ="New York")
```

Global Variable 

```python
# A variable defined outside the function that is known as global variable.
# That can acess all Fuction


a= 10 #global variable
def Show():
    a=10
    print("global",a)
def disp():
    print(a)
    print("hii")
show()
disp()
print(a)
```
