```python
# create a fun to check no is perfect or not(5,6,49,28,67,113)

def is_armstrong(num):
    power = len(str(num))
    temp = num
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** power
        temp //= 10

    return total == num

numbers = [9, 153, 2248, 487, 1153]

for n in numbers:
    if is_armstrong(n):
        print(n, "is an Armstrong number ")
    else:
        print(n, "is NOT an Armstrong number ")
```
