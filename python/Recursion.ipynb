```python
def facttest(no):
    f=1
    while  no>0:
        f=f*no
        no=no-1
    return f
res=facttest(3)
print(res)
```

```python
# sum of digit using funcction
def adtest(no):
    s=0
    while no!=0:
        r = no%10
        s=s+r
        no=no//10
    return s
res = adtest(123)
print(res)
```

```python
# using recursion
s=0
def adtest(no):
    global s 
    if no!=0:
        r = no%10
        s=s+r
        no=no//10
        adtest(no)
    return s
res = adtest(123)
print(res)
```
