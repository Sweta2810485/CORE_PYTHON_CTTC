```python
# Factorial of a number

n = int(input("Enter a number : "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
    print(factorial)
```

```python
# Product of number.

n = int(input("Enter a number : "))
product = 1
for digit in str(n):
    product *= int(digit)
    print(product)
```

```python
st = input("enter a string:")
count_num = 0
count_alpha = 0
for i in st:
    if i.isdigit():
        count_num +=1
    if i.isalpha():
        count_alpha += 1
print(f"The number of alphabet{count_alpha} and number of digit {count_num}")
```

```python
# Wapp to print count the factors of an entered number.


number = int(input("Enter a number"))
count = 0
for i in range(1,number+1):
    if number % i == 0:
        count += 1
print(f"count of factor numbers: {count}")
```

```python
# Wapp that accepts a word from the user and reverses it.
w = input("Enter a word: ")
reverse_w = ""

for i in range(len(w) - 1, -1, -1):
    reverse_w += w[i]
print(f"Reversed word:{reverse_w}")
```

```python
# Wapp to check if the entered string is palindrome or not.


s = input("Enter a string: ")
reversed_s = ""
for char in s:
    reversed_s = char + reversed_s
if s== reversed_s:
    print("It is a palindrome")
else:
    print("Not a palindrome")
```

```python
 # Wapp to check if the entered number is palindrome or not.


num = input("Enter the number: ")
reversed_num = ""
for digit in num:
    reversed_num = digit + reversed_num
if num == reversed_num:
    print("palindrome number.....")
else:
    print("Chala ja bsdk its not a palindrome number")
```
