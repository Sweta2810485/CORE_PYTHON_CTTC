# """Use constructor and check if a given number is prime or not.
# Take a class prime , In the Instance variable take that number.
# In the calculation method check the number prime or not print."""


```python


class prime:
    def __init__(self,number):
        self.number= number
    def is_prime(self):
        if self.number < 2:
            return False
        for i in range(2, int(self.number ** 0.5) + 1):
            if self.number % i == 0:
                return False
        return True
        
        
```

```python
num = prime(7)
num.is_prime()
```
