# Inheritance 


Inheritance is a concept to inherit values and method from one class to another class.

single inheritance
In single inheritance there is one child clss and one parent class.


```python
#ex-
class father:
    f_p="31 ecr"
class mother:
    m_p="31 ecr"
class son(father,mother):
    s_p="1 erc"
    
```

```python
s=son()
print(s.s_p)
print(s.f_p)
print(s.m_p)
```

#miltiple inheritance = miltiple inheritance there can be multiple parent class for one child class

```python
class father:
    f_p="31 ecr"
class mother:
    m_p="31 ecr"
class son(father,mother):
    s_p="1 erc"
    
```

```python
s=son()
print(s.s_p)
print(s.f_p)
print(s.m_p)
```

```python
#multilevel inheritance= in multilevel inheritance a child class inhert it property from another child class which inherit the propert from one child class to another class. 
```

```python
class grandfather:
    g_p="31 ecr"
class father(grandfather):
    f_p="31 ecr"
class son(father):
    s_p="1 erc"
    
```

```python
s=son
print(s.s_p)
print(s.f_p)
print(s.g_p)
```

```python
heirechical class=in heirechical inhreitance there is one parent class to multiple child class
```

```python
#heirechical inhreitance 
class grandfather:
    g_p="31 ecr"
class father(grandfather):
    f_p="31 ecr"
class son(father):
    s_p="1 erc"
class son2(father):
    s2_p="1 erc"
```

```python
s=son()
print(s.s_p)
print(s.f_p)
s5=son2()
print(s5.s2_p)
print(s5.f_p)
```

```python
#hybrid inheritance = mixture of two or more  inheritances
```

```python
class grandfather:
    g_p="31 ecr"
class father(grandfather):
    f_p="31 ecr"
class son(father):
    s_p="1 erc"
class son2(father):
    s2_p="1 erc"
```

```python
s=son()
print(s.s_p)
print(s.f_p)
print(s.g_p)
s5=son2()
print(s5.s2_p)
print(s5.f_p)
print(s5.g_p)
```
