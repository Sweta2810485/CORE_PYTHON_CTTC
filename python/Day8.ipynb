```python
# Check the number is ARMSTRONG or not using combine for and while loop.

num = int(input("Enter a number: "))
sum = 0
temp = num

count = 0
while temp > 0:
    count += 1
    temp //= 10

temp = num

while temp > 0:
    digit = temp % 10
    for _ in [digit]:  
        sum += digit ** count
    temp //= 1015
    
if sum == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is NOT an Armstrong number")
```
