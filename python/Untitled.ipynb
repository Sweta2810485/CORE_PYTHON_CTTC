Constructor:

It is a special method.
It looks like __init__(self,...) other language same as class name(java,c,c++)
when object is created constructor excute.
3 object 3 times constructor excute automatically.
It is used to initlize instance variable.

```python
class demo:
    def __init__(self):
        print("Constructor")
d = demo()
d1 = demo()
```

```python
Destructor:

It is a special method.
When object destory automatic call.
```

```python
class demo:
    def __init__(self):
        print("Constructor")
    def __del__(self):
        print("Destructor")
d = demo()
d1 = demo()
d2= demo()
```

(2) Class Method:

It is related to class.
We create class method by using @classmethod decorator
It is call by classname.not required object
It takes one argument that acess class variable

```python
# Class method:(Call class)
class Demo:
    @classmethod
    def disp(cls):
        print("Disp class method")
Demo.disp()
```

(3) Static method:

It is related to class.
We create class method by using @classmethod decorator
It is call by classname.Not required object
It takes on argument
using static method we create utlity class

```python
class Demo:
    @staticmethod
    def disp():
        print("disp static method")
Demo.disp()
d1=Demo()
d1.disp()
        
```

Class variable:
A variable create inside class
inside classmethod using cls variable

Outside class by using classname.

```python


class demo():
    a=10
demo.b=20
print(demo.__dict__)
```

```python
class demo():
    a=10
    @classmethod
    def set(cls):
        cls.c=30
demo.b=20
demo.set()
print(demo.__dict__)
```

How to delete class variable.

```python
class demo():
    a=10
    @classmethod
    def set(cls):
        cls.c=30 # Delete by cls
        del cls.b
demo.b=20
demo.set()
del demo.a # delete by class name
print(demo.__dict__)
```
