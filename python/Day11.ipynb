```python
def fibo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibo(n-1) + fibo(n-2)
```

```python
n = 5
for i in range(n):
    print(fibo(i),end = " ")
```

```python
# lambda
fibona = lambda n : n if n<=1 else fibo(n-1) + fibo(n-2)
```

```python
n = 5
for i in range(n):
    print(fibo(i),end = " ")
```

```python
list(map(fibona,[5,6]))
```

```python
# reverse a number using lambda
def reverse_number (n, rev = 0):
    if n == 0:
        return rev
    else:
        return reverse_number(n//10,rev *10 + n %10) 
    
```

```python
reverse_number(123)
```

```python
reverse = lambda n,rev = 0 : rev if n== rev else reverse_number(n//10,rev *10 + n %10) 
```

```python
reverse(123)
```

```python
# sum of digit of a number using lambda
def sum_digits(n):
    if n ==0:
        return  0
    else:
        return n%10 + sum_digits(n//10)
        

```

```python
sum_digits(123)
```

# filter function

return all the element of the sequence that satisfy the function 

```python
even = lambda n : True if n%2==0 else False
list(filter(even,[45,536,342,32]))
```
