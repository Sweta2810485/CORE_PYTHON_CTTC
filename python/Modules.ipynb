```python
# It is a python file.
# Inside module we write Function variable Class.
# There are 2 types of modules.
# 1 - Predefined module
# sys
# random
# numpy
# panda


# 2 -Userdefined module
# That name must be valid identifier


# How to use module or access module

# We can use module 4 Way
# 1- import modulename
# 2- import modulename as dupname
# 3- from modulename import member, member
# 4- from modulename import



mymodule.py

def show():
    print("Show function.")
def add(no1,no2):
    return no1+no2
a =10


first.py

import mymodule
mymodule.show()
print("sum=",mymodule.add(10,20))
print(mymodule.a)

second.py
import mymodule as m
m.show()
print("sum=",m.add(10,20))
print(m,a)

third.py
from mymodule import show,add,a
show()
print("sum=",add(10,20))
print(a)



four.py
from mymodule import *
show()
print("sum=",add(10,20))
print(a)


```

CHOICE

```python
import random
print(random.choice("Welcome"))
print(random.choice([11,2,3,4,3]))
print(random.choice(range(10)))
      
```

Sample

```python
import random
print(random.sample("Welcome",3))
print(random.sample([11,2,3,4,3],4))
print(random.sample(range(10),2))
      
```
